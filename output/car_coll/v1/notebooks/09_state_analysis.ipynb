```python
with open(f'{config_path}/config.yaml') as f:
    cfg = yaml.safe_load(f)
# Get current machine ID and output path
pc_num = open('current.pc').read().strip()
pc_id = f'PC{pc_num}'  # current.pc has '3', config key is 'PC3'
output_base = cfg['machines'][pc_id]['paths']['output_path']
analysis_cfg = cfg['analysis']
```

```python
def plot_lift(df, target, pred, weight, file, title):
    df = df.copy()
    df['d'] = pd.qcut(df[pred], 10, labels=False, duplicates='drop')
    s = df.groupby('d').agg({target: 'sum', pred: 'sum', weight: 'sum'}).reset_index()
    s['ar'] = s[target] / s[target].sum()
    s['pr'] = s[pred] / s[pred].sum()
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(s['d'], s['ar'], marker='o', label='Actual')
    ax.plot(s['d'], s['pr'], marker='s', label='Predicted')
    ax.set_xlabel('Decile')
    ax.set_ylabel('Relative')
    ax.set_title(title)
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(file, dpi=150, bbox_inches='tight')
    plt.close()
```
