```python
config_path = "config/car_coll/v1/config.yaml"
```

```python
import os
import yaml
import pandas as pd

with open(config_path, 'r') as f:
    cfg = yaml.safe_load(f)

config_dir = os.path.dirname(config_path)
print(f"Config: {config_path}")
```

```python
veh_type = cfg['experiment']['veh_type']
path_prefix = cfg['paths']['path_prefix']
in_data_store = path_prefix + cfg['paths']['input_data_dir']
input_file = cfg['files']['input_file_pattern'].format(veh_type=veh_type)
input_path = in_data_store + input_file

print(f"Reading: {input_path}")
```

```python
# Read first 1000 rows to get column info
data = pd.read_csv(input_path, nrows=1000)
print(f"Columns: {len(data.columns)}")
```

```python
# Extract feature columns (vc_ prefix)
feature_cols = [c for c in data.columns if c.startswith('vc_')]
print(f"Feature columns: {len(feature_cols)}")
```

```python
# Create all_features.csv
features_df = pd.DataFrame({'feature_name': feature_cols})
features_df['selected'] = 0
features_df['blank'] = 0

output_path = os.path.join(config_dir, 'all_features.csv')
features_df.to_csv(output_path, index=False)
print(f"Saved: {output_path}")
print(f"Total features: {len(features_df)}")
```
